# Análisis estadístico de la simulación secuencial de nanodiscos con DMI

Ocho estados (lotes 1–4), escalera de 20 K a 6 K con paso 0.1 K, 100 copias independientes y 10 000 mediciones por temperatura.

- **Etapa A** (1128 tareas independientes, en paralelo): verificación de cada archivo, sumas por copia, τ_int de Sokal, jackknife por bloques, diagnósticos de convergencia (Geweke, MSER-5, R̂ por rangos, ESS, homogeneidad entre copias), repesado de histograma a una rejilla de 0.01 K y análisis de las configuraciones de espines (carga topológica, solapamiento, factor de estructura).
- **Etapa B** (un proceso por estado): bootstrap entre copias con B = 50 000 (la copia completa es la unidad de remuestreo, con las mismas multiplicidades en toda la escalera), intervalos percentil y BCa, distribución de T*.
- **Etapa C**: comparación entre estados, tablas, figuras y veredictos.

Las etapas A y B se ejecutan con `ejecutar.py` (reanudable); este cuaderno las relanza (lo hecho se salta) y presenta los resultados.
